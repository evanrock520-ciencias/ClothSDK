```python
from tissu import engine
```

```python
sim = engine.Simulation(substeps=15, iterations=3, thickness=0.05)
```

```python
dress = sim.create_from_obj(
    name="dress",
    path="assets/models/dress.obj",
    material="silk"
)

dress.pin_top_corners()
```
