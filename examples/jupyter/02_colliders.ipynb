```python
from tissu import engine
```

```python
sim = engine.Simulation(thickness=0.05, gravity=-20.0)

curtain = sim.create_grid(
    name="curtain", rows=80, cols=80, spacing=0.05, material="silk"
)

curtain.pin_top_corners()
```

## Add Collider

```python
sim.add_sphere(name="sphere", center=[2.0, 2.0, -0.5], radius=1.0)
```

## Add Callback

```python
@sim.on_frame(40)
def unpin(sim):
    curtain.unpin()
```

## Render Gif

```python
sim.plot_gif("curtain", filename="out/gif/collider.gif")
```
