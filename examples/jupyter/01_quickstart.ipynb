## Importation

```python
from tissu import engine 
```

```python
sim = engine.Simulation(substeps=8, thickness=0.05, gravity=-20.0)

# Create fabric
curtain = sim.create_grid(
    name="curtain",
    rows=80, 
    cols=80,
    spacing=0.05, # Same as simulation's thickness
    material= "silk"
)
```

## Pin Curtain

```python
curtain.pin_top_corners()
```

```python
sim.simulate(160)
```

## Preview

```python
sim.plot("curtain")
```
