```python
from tissu import Simulation
```

```python
sim = Simulation(substeps=10, iterations=3, thickness=0.05)
curtain = sim.create_grid(
    name="curtain",
    rows=80,
    cols=80,
    spacing=0.05,
    material="silk"
)
```

```python
initial = -2.0
speed = 0.05
sim.add_sphere("ball", center=[initial, 0.0, 0.0], radius=1, friction=0.5)
```

```python
@sim.on_range(0, 180)
def move_ball(sim):
    t = sim.frame / 180.0
    x = -0.8 + t * 1.6
    sim.move_collider("ball", [x, 0.0, 0.0])
    
```

# Previsualize

```python
sim.plot("curtain")
```

# Simulate

```python
sim.start_recording()
sim.simulate(120)
sim.stop_recording()
```
