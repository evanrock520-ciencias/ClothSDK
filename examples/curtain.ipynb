```python
from tissu import Simulation
```

```python
sim = Simulation(substeps=10, iterations=3, thickness=0.05)
curtain = sim.create_grid(
    name="curtain",
    rows=80,
    cols=80,
    spacing=0.05,
    material="silk"
)
```

```python
sim.add_capsule(
    start=(0.0, 0.0, 0.0),
    end=(0.0, -1.0, 0.0),
    radius=1.0
    )
```

# Previsualize

```python
sim.view()
```

# Bake

```python
sim.bake_alembic(
    filepath="../data/animations/bounce.abc",
    start_frame=0,
    end_frame=180,
    fps=60
)
```
